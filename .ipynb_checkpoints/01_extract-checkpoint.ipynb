```python
from dotenv import load_dotenv
import os
import requests
from sqlalchemy import create_engine
import pandas as pd

load_dotenv()
engine = create_engine(os.getenv("DATABASE_URL"))

# Fetch top 50 scorers instead of default 10
headers = {"X-Auth-Token": os.getenv("FOOTBALL_API_KEY")}
url = "https://api.football-data.org/v4/competitions/PL/scorers?limit=50"
response = requests.get(url, headers=headers)
data = response.json()

rows = []
for entry in data["scorers"]:
    player = entry["player"]
    team   = entry["team"]
    rows.append({
        "scorer_id":      player["id"],
        "player_name":    player["name"],
        "team_id":        team["id"],
        "team_name":      team["name"],
        "played_matches": entry.get("playedMatches", 0),
        "goals":          entry.get("goals", 0),
        "assists":        entry.get("assists") or 0,
        "penalties":      entry.get("penalties") or 0,
    })

df = pd.DataFrame(rows)
df["goals_per_match"] = (df["goals"] / df["played_matches"].replace(0, pd.NA)).round(2)

# Replace scorers table with expanded data
df.to_sql("scorers", engine, schema="public", if_exists="replace", index=False)
print(f"Scorers updated: {len(df)} players loaded")
print(df[["player_name", "team_name", "goals"]].head(20))
```

```python
from dotenv import load_dotenv
import os
import requests
from sqlalchemy import create_engine, text
import pandas as pd

load_dotenv()
engine = create_engine(os.getenv("DATABASE_URL"))

# Fetch top 100 scorers
headers = {"X-Auth-Token": os.getenv("FOOTBALL_API_KEY")}
url = "https://api.football-data.org/v4/competitions/PL/scorers?limit=150"
response = requests.get(url, headers=headers)
data = response.json()

rows = []
for entry in data["scorers"]:
    player = entry["player"]
    team   = entry["team"]
    rows.append({
        "scorer_id":      player["id"],
        "player_name":    player["name"],
        "team_id":        team["id"],
        "team_name":      team["name"],
        "played_matches": entry.get("playedMatches", 0),
        "goals":          entry.get("goals", 0),
        "assists":        entry.get("assists") or 0,
        "penalties":      entry.get("penalties") or 0,
    })

df = pd.DataFrame(rows)
df["goals_per_match"] = (df["goals"] / df["played_matches"].replace(0, pd.NA)).round(2)

# Check Wolves
wolves = df[df["team_name"].str.contains("Wolverhampton")]
print(f"Wolverhampton scorers: {len(wolves)}")
print(wolves[["player_name", "goals"]])

# Show all teams represented
print(f"\nTotal scorers: {len(df)}")
print(f"Teams covered: {df['team_name'].nunique()}")
print(df.groupby('team_name')['player_name'].count().sort_values())
```

```python
df.to_sql("scorers", engine, schema="public", if_exists="replace", index=False)
print(f"Database updated: {len(df)} scorers loaded")
```
